```python
# create model class
import torch
import torch.nn as nn

class Model(nn.Module):# always use nn.modules

  def __init__(self, num_features):

    super().__init__()
    self.network = nn.Sequential(
        nn.Linear(num_features, 3),#(input features, output features)
        nn.ReLU(),#actuvation fn
        nn.Linear(3, 1),
        nn.Sigmoid() #activation fn
    )
# this is called sequential containers
  def forward(self, features):

    out = self.network(features)

    return out
```

```python
# create dataset
features = torch.rand(10,5)

# create model
model = Model(features.shape[1])

# call model for forward pass
# model.forward(features)
model(features) #standard
```

```python
# show model weights
model.linear.weight
```
